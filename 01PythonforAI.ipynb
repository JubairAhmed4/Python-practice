```python
revenue =[60,40,20]
expense=[30,33,36]

print(len(revenue))
print(len(expense))
print(revenue[0:])
print(expense[0:])


for i in range(len(revenue)):
    profit = revenue[i] - expense[i]
    margin = profit*100 / revenue[i]
    print(f"Margin : {margin:.3f}")
```
